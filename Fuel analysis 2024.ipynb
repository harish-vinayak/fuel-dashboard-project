```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Engine size (L)', y='CO2 emissions (g/km)')
plt.title('Engine Size vs CO2 Emissions')
plt.xlabel('Engine Size (L)')
plt.ylabel('CO2 Emissions (g/km)')
plt.grid(True)
plt.show()
```

```python
correlation_matrix = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

```python
fuel_emissions = df.groupby('Fuel Type')['CO2 emissions (g/km)'].mean().sort_values()
fuel_emissions.plot(kind='bar', figsize=(8, 5), color='green')
plt.title('Average CO2 Emissions by Fuel Type')
plt.ylabel('CO2 Emissions (g/km)')
plt.xlabel('Fuel Type')
plt.grid(axis='y')
plt.show()
```

```python
engine_size = fuel_ratings_imputed['Engine size (L)']
cylinders = fuel_ratings_imputed['Cylinders']
fuel_consumption = fuel_ratings_imputed['Combined (L/100 km)']
comb_rate = fuel_ratings_imputed['Combined (mpg)']
smog_rating = fuel_ratings_imputed['Smog rating']

data_to_plot = [engine_size, cylinders, fuel_consumption, comb_rate, smog_rating]

plt.figure(figsize=(10, 8))
plt.boxplot(data_to_plot, tick_labels=['Engine Size', 'Cylinders', 'Fuel Consumption', 'Comb rate', 'Smog Rating'])
plt.show()
```
